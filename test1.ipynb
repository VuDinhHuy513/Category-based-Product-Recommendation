# Đọc dữ liệu

```python
import pandas as pd

df = pd.read_csv("amazon.csv")
print(df.shape)
print(df.columns)
df.head()
```

# Chọn những cột cần dùng

```python
cols_needed = [
    "product_id",
    "product_name",
    "category",
    "discounted_price",
    "actual_price",
    "discount_percentage",
    "rating",
    "rating_count",
    "about_product",
    "img_link",
    "product_link"
]

df = df[cols_needed].copy()
```

# Ép kiểu về string và loại bỏ khoảng trắng

```python
text_cols = ["product_id", "product_name", "category", "about_product", "img_link", "product_link"]

for col in text_cols:
    df[col] = df[col].astype(str).str.strip()
```

# Hàm sử lí dữ liệu số

```python
import numpy as np
import re

def clean_price(x):
    if pd.isna(x):
        return np.nan
    x = str(x).replace("₹", "").replace(",", "").strip()
    try:
        return float(x)
    except:
        return np.nan

def clean_percentage(x):
    if pd.isna(x):
        return np.nan
    x = str(x).replace("%", "").strip()
    try:
        return float(x)
    except:
        return np.nan

def clean_count(x):
    if pd.isna(x):
        return np.nan
    x = str(x).replace(",", "").strip()
    try:
        return int(float(x))
    except:
        return np.nan

def clean_rating(x):
    if pd.isna(x):
        return np.nan
    x = str(x).strip()
    try:
        return float(x)
    except:
        return np.nan
```

```python
df["discounted_price"] = df["discounted_price"].apply(clean_price)
df["actual_price"] = df["actual_price"].apply(clean_price)
df["discount_percentage"] = df["discount_percentage"].apply(clean_percentage)
df["rating"] = df["rating"].apply(clean_rating)
df["rating_count"] = df["rating_count"].apply(clean_count)
```

```python
df = df[(df["rating"].isna()) | ((df["rating"] >= 0) & (df["rating"] <= 5))]
df = df[(df["discounted_price"].isna()) | (df["discounted_price"] >= 0)]
df = df[(df["actual_price"].isna()) | (df["actual_price"] >= 0)]
df = df[(df["rating_count"].isna()) | (df["rating_count"] >= 0)]
```

# Xử lí missing

```python
df["product_name"] = df["product_name"].fillna("")
df["about_product"] = df["about_product"].fillna("")
df["category"] = df["category"].fillna("Unknown")
df["rating"] = df["rating"].fillna(df["rating"].median())
df["rating_count"] = df["rating_count"].fillna(0)
```

# Xử lí dữ liệu trùng

```python
agg_dict = {
    "product_name": "first",
    "category": "first",
    "discounted_price": "first",
    "actual_price": "first",
    "discount_percentage": "first",
    "rating": "mean",
    "rating_count": "max",
    "about_product": "first",
    "img_link": "first",
    "product_link": "first"
}

df = df.groupby("product_id", as_index=False).agg(agg_dict)
```

# Phân cấp category

```python
cat_split = df["category"].str.split("|", expand=True)

max_levels = cat_split.shape[1]
cat_split.columns = [f"cat_level_{i+1}" for i in range(max_levels)]

df = pd.concat([df, cat_split], axis=1)
df = df.rename(columns={"category": "category_path"})
```

# Điền dữ liệu thiếu cho các cấp category

```python
for col in [c for c in df.columns if c.startswith("cat_level_")]:
    df[col] = df[col].fillna("Unknown")
```

# làm sạch text

```python
def clean_text(text):
    text = str(text).lower()
    text = text.replace("&", " and ")
    text = text.replace("|", " ")
    text = re.sub(r"[^a-z0-9\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text
```

```python
df["product_name_clean"] = df["product_name"].apply(clean_text)
df["about_product_clean"] = df["about_product"].apply(clean_text)
df["category_text"] = df["category_path"].apply(clean_text)
```

```python
df["clean_text_no_category"] = (
    df["product_name_clean"] + " " + df["about_product_clean"]
)
```

```python
df["clean_text_with_category"] = (
    df["product_name_clean"] + " " +
    df["about_product_clean"] + " " +
    df["category_text"]
)
```

```python
df = df[df["product_name_clean"].str.len() > 5]
df = df[df["category_text"].str.len() > 2]
```

```python
print(df.shape)
print(df.isna().sum())
df.head()
```

```python
df.to_csv("clean_products.csv", index=False, encoding="utf-8-sig")
```
