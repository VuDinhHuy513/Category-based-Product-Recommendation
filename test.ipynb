# Đọc dữ liệu

```python
path='amazon.csv'
```

```python
df = pd.read_csv(path)
df.head()
```

```python
df.info()
```

```python
df = df.drop_duplicates(subset="product_id")
df = df.reset_index(drop=True)
df.info()
```

```python
df["product_name"]
```

# Lấy feature

```python
df = df[["product_id","product_name","category","about_product","rating"]]
```

# Tạo feature text

```python
df["combined_text"] = (df["product_name"] + " " + df["about_product"] + " " + df["category"])
```

```python
tfidf = TfidfVectorizer(stop_words="english")
tfidf_matrix = tfidf.fit_transform(df["combined_text"])
tfidf_matrix.shape
```

```python
similarity_matrix = cosine_similarity(tfidf_matrix)
```

```python
def recommend(product_name, top_n=5):

    idx = df[df["product_name"] == product_name].index[0]

    sim_scores = list(enumerate(similarity_matrix[idx]))

    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)

    sim_scores = sim_scores[1:top_n+1]

    product_indices = [i[0] for i in sim_scores]

    return df.iloc[product_indices][["product_name","category","rating"]]
```

```python
recommend("Wayona Nylon Braided USB to Lightning Fast Charging and Data Sync Cable Compatible for iPhone 13, 12,11, X, 8, 7, 6, 5, iPad Air, Pro, Mini (3 FT Pack of 1, Grey)")
```
