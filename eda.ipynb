```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Đọc dữ liệu sạch
df = pd.read_csv("data/processed/clean_products.csv")

# Kiểm tra nhanh dữ liệu
print(df.shape)
print(df.columns)
df.head()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="rating",
    bins=20,
    kde=True
)

plt.title("Distribution of Product Ratings", fontsize=14)
plt.xlabel("Rating")
plt.ylabel("Number of Products")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="rating",
    order=sorted(df["rating"].dropna().unique())
)

plt.title("Product Rating Distribution", fontsize=14)
plt.xlabel("Rating")
plt.ylabel("Number of Products")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="rating_count",
    y="rating",
    alpha=0.6
)

plt.title("Relationship between Rating Count and Rating", fontsize=14)
plt.xlabel("Rating Count")
plt.ylabel("Rating")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="rating_count",
    y="rating",
    alpha=0.6
)

plt.xscale("log")

plt.title("Relationship between Rating Count and Rating (Log Scale)", fontsize=14)
plt.xlabel("Rating Count (Log Scale)")
plt.ylabel("Rating")

plt.tight_layout()
plt.show()
```

```python
top_cat1 = df["cat_level_1"].value_counts().head(10)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=top_cat1.values,
    y=top_cat1.index
)

plt.title("Top Category Level 1 by Number of Products", fontsize=14)
plt.xlabel("Number of Products")
plt.ylabel("Category Level 1")

plt.tight_layout()
plt.show()
```

```python
top_cat3 = df["cat_level_3"].value_counts().head(10)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_cat3.values,
    y=top_cat3.index
)

plt.title("Top Category Level 3 by Number of Prducts", fontsize=14)
plt.xlabel("Number of Products")
plt.ylabel("Category Level 3")

plt.tight_layout()
plt.show()
```
